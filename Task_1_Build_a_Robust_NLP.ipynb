# Build Advanced Preprocessing Function

```python
import re

def preprocess_text(text):
    """
    Cleans and preprocesses input text and returns tokens + cleaned sentence
    """

    #  Error Handling
    if not isinstance(text, str) or text.strip() == "":
        return [], ""

    # 1 Convert to lowercase
    text = text.lower()

    # 2️ Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)

    # 3️ Remove email-like patterns
    text = re.sub(r"\S+@\S+", "", text)

    # 4️ Remove numbers
    text = re.sub(r"\d+", "", text)

    # 5️ Handle repeated characters (soooo → so)
    text = re.sub(r"(.)\1{2,}", r"\1", text)

    # 6️ Remove emojis & special characters
    text = re.sub(r"[^a-z\s]", "", text)

    # 7️ Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # 8️ Tokenization
    tokens = text.split()

    # 9️ Remove very short words (≤2) except 'no', 'not'
    tokens = [word for word in tokens if len(word) > 2 or word in ["no", "not"]]

    #  Create cleaned sentence
    cleaned_sentence = " ".join(tokens)

    return tokens, cleaned_sentence
```

```python
text = "I absolutely looooved this product "

tokens, cleaned = preprocess_text(text)

print("Tokens:", tokens)
print("Cleaned Sentence:", cleaned)
```

# Stress Testing


```python
test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product ",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]
```

```python
results = []

for sentence in test_sentences:
    tokens, cleaned = preprocess_text(sentence)

    results.append({
        "original": sentence,
        "tokens": tokens,
        "cleaned": cleaned
    })
```

```python
for r in results:
    print("\n-----------------------------")
    print("Original Text   :", r["original"])
    print("Cleaned Tokens  :", r["tokens"])
    print("Cleaned Sentence:", r["cleaned"])
```

# Token Analytics


```python
import numpy as np

analytics = []

for r in results:
    tokens = r["tokens"]

    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))

    # Handle empty tokens safely
    avg_length = np.mean([len(word) for word in tokens]) if tokens else 0

    analytics.append({
        "sentence": r["original"],
        "total_tokens": total_tokens,
        "unique_tokens": unique_tokens,
        "avg_length": round(avg_length, 2)
    })
```

```python
for a in analytics:
    print("\n-----------------------------")
    print("Sentence        :", a["sentence"])
    print("Total Tokens    :", a["total_tokens"])
    print("Unique Tokens   :", a["unique_tokens"])
    print("Avg Token Length:", a["avg_length"])
```

# Build Full Pipeline


```python
def full_pipeline(text_list):
    """
    Processes a list of sentences and returns tokens + cleaned sentences
    """

    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, cleaned = preprocess_text(text)

        all_tokens.extend(tokens)          # Combine all tokens
        clean_sentences.append(cleaned)    # Store cleaned sentence

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
```

```python
output = full_pipeline(test_sentences)

print("All Tokens:\n", output["tokens"])
print("\nCleaned Sentences:\n", output["clean_sentences"])
```

# Error Handling


```python
import re

def preprocess_text(text):
    """
    Robust preprocessing function with error handling
    """

    # 1️ Handle invalid or empty input
    if not isinstance(text, str) or text.strip() == "":
        return [], ""

    # 2️ Convert to lowercase
    text = text.lower()

    # 3️ Remove URLs & emails
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)
    text = re.sub(r"\S+@\S+", "", text)

    # 4️ Remove numbers
    text = re.sub(r"\d+", "", text)

    # 5️ Handle repeated characters
    text = re.sub(r"(.)\1{2,}", r"\1", text)

    # 6️ Remove emojis & special characters
    text = re.sub(r"[^a-z\s]", "", text)

    # 7️ Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # 8️ Tokenization
    tokens = text.split()

    # 9️ Remove short words (≤2) except 'no', 'not'
    tokens = [word for word in tokens if len(word) > 2 or word in ["no", "not"]]

    #  Final cleaned sentence
    cleaned_sentence = " ".join(tokens)

    return tokens, cleaned_sentence
```

```python
test_cases = [
    "",                     # Empty string
    "😍🔥😂",              # Only emojis
    "1234567890",          # Only numbers
    "     ",               # Only spaces
    None                   # Invalid input
]

for t in test_cases:
    tokens, cleaned = preprocess_text(t)
    print("\nInput:", t)
    print("Tokens:", tokens)
    print("Cleaned:", cleaned)
```
